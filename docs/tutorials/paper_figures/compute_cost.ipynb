# Supplementary Table 2: full-model compute cost

Load the packaged measurements, apply manuscript units, and save the table.


## Load


````python
from pathlib import Path

from IPython.display import display

from CytoBridge.results.compute_cost import (
    format_full_model_compute_cost,
    load_full_model_compute_cost,
    write_full_model_compute_cost_tables,
)

results = load_full_model_compute_cost()
display(results.measurements)
````

## Format

Convert seconds to minutes and MiB to GiB.


````python
table = format_full_model_compute_cost(results)
display(table)
````

## Save and check


````python
output_dir = Path("outputs") / "full_model_compute_cost_notebook"
paths = write_full_model_compute_cost_tables(results, output_dir)
{name: {"file": path.name, "written": path.is_file()} for name, path in paths.items()}
````
